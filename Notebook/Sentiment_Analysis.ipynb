```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('amazon_electronics_sample.csv')
```

```python
df.head()
```

```python
# Rename columns for better readability
df.rename(columns={
    'reviewerID': 'reviewer_id',
    'reviewerName': 'reviewer_name',
    'reviewText': 'review_text',
    'reviewTime': 'review_time',
    'unixReviewTime': 'unix_review_timestamp',
    'overall': 'overall_rating',
    'helpful': 'helpful_votes',
    'summary': 'review_summary',
    'asin': 'product_id'
}, inplace=True)
```

```python
df.head()
```

```python
df['review_text'] = df['review_text'].str.lower()
df.head(10)
```

```python
import re
df['review_text'] = df['review_text'].apply( lambda x: re.sub(r'[^\w\s]|\d+', '', str(x))) 
df.head(10)
```

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('English'))

negative_words = {'no','not','never','neither','nor'}

stop_words =  stop_words - negative_words

df['review_text'] = df['review_text'].apply(
    lambda text: ' '.join(
        word for word in text.split()
        if word not in stop_words
    )
)
df.head()
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize

df['review_text'] = df['review_text'].apply(word_tokenize)
df.head()
```

```python
from nltk.stem import WordNetLemmatizer
import nltk

nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()

df['review_text'] = df['review_text'].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)

df[['review_text']].head()
```

```python
type(df['review_text'].iloc[0])
```

```python
df['review_text'] = df['review_text'].apply(lambda words: ' '.join(words))

df['review_text'].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(df['review_text'])
```

```python
X[:5].toarray()

vectorizer.get_feature_names_out()

len(vectorizer.get_feature_names_out())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(df['review_text'])
X_tfidf
```

```python
X_tfidf[:5].toarray()

tfidf.get_feature_names_out()

len(tfidf.get_feature_names_out())
```

```python
def get_sentiments(rating):
    if rating < 3:
        return 'negative'
    elif rating == 3:
        return 'neutral'
    else:
        return 'positive'

df['Sentiment'] = df['overall_rating'].apply(get_sentiments)
```

```python
df.head()
```

```python
sentiment_counts = df['Sentiment'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    sentiment_counts,
    labels=sentiment_counts.index,
    autopct ='%1.1f%%',
    startangle = 90
)

plt.title('Distribution of Customer Sentiments')
plt.savefig("Customer_sentiments_Pie.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
from sklearn.model_selection import train_test_split

x = X_tfidf
y = df['Sentiment']

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size = 0.2,
    random_state = 42
)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state = 42)

model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)

print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print(f'Accuracy : {accuracy}')
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
positive_reviews = df[df['Sentiment'] == 'positive']['review_text']

print(positive_reviews.head())
```

```python
from collections import Counter

positive_text = " ".join(positive_reviews)

word_count = Counter(positive_text.split())

print(word_count.most_common(20))
```

```python
custom_stopwords = {"one", "use", "get", "would", "also","like", "well", "time", "work" }

filtered_words = {
    word : count
    for word, count in word_count.items()
    if word not in custom_stopwords and len(word) > 2
}

filtered_counts = Counter(filtered_words)

print(filtered_counts.most_common(20))
```

```python
negative_reviews=  df[df['Sentiment'] == 'negative']['review_text']

negative_text = " ".join(negative_reviews)

negative_word_count = Counter(negative_text.split())

filtered_negative_words = {
    word: count
    for word, count in negative_word_count.items()
    if word not in custom_stopwords and len(word) > 2
}

filtered_negative_counts = Counter(filtered_negative_words)

print(filtered_negative_counts.most_common(20))
```

```python
top10_positive = {
    (word, count)
    for word, count in filtered_negative_counts.most_common(10)
    if word != 'not'
    }
word = [word for word, count in top10_positive]
count = [count for word, count in top10_positive]

plt.figure(figsize = (10,6))
plt.bar(word, count)
for i, value in enumerate(count):
    plt.text(i, value, str(value), ha='center', va='bottom')
plt.title('Top 10 Most Frequent Positive words', fontsize = 18)
plt.xlabel('Words', fontsize = 16)
plt.ylabel('Frequency', fontsize = 16)
plt.tight_layout()
plt.savefig("Top_10_Positive_Words.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
top10_negative = filtered_negative_counts.most_common(10)

word = [word for word,count in top10_negative]
count = [count for word,count in top10_negative]

plt.figure(figsize = (10,6))
plt.bar(word,count)

for i, value in enumerate(count):
    plt.text(i, value, str(value), ha='center', va='bottom')
    
plt.title('Top 10 Most Frequent Negative Words', fontsize = 18)
plt.xlabel('Words', fontsize = 16)
plt.ylabel('Frequency', fontsize = 16)
plt.tight_layout()
plt.savefig("Top_10_Negative_Words.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
from wordcloud import WordCloud, STOPWORDS

stopwords = STOPWORDS.union(custom_stopwords)

stopwords.update({
    "used",
    "using",
    "product",
    "really",
    "much",
    "even",
    "well",
    "good",
    "great",
    "need",
    "dont"
})

stopwords.update({
    "bought",
    "thing",
    "make",
    "first",
    "better",
    "see",
    "set",
    "want",
    "come",
    "look"
})
```

```python
positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width = 800,
    height= 400,
    background_color = 'white',
    stopwords = stopwords
).generate(positive_text)

plt.figure(figsize = (12,6))
plt.imshow(wordcloud)
plt.axis('off')
plt.title('Word cloud of positive reviews', fontsize = 20, fontweight = 'bold')
plt.savefig("WordCloud_Positive.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python
negative_text = " ".join(negative_reviews)

wordcloud = WordCloud(
    width = 800,
    height= 400,
    background_color = 'white',
    stopwords = stopwords
).generate(negative_text)

plt.figure(figsize = (12,6))
plt.imshow(wordcloud)
plt.axis('off')
plt.title('Word cloud of negative reviews', fontsize = 20, fontweight = 'bold')
plt.savefig("WordCloud_negative.png",
            dpi=300,
            bbox_inches="tight")
plt.savefig("WordCloud_Positive.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```
